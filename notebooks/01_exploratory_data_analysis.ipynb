```python
import pandas as pd
import os

# 1. Load the dataset path
# '..' means go up out of notebooks folder, then look into data -> raw
data_path = os.path.join("..", "data", "raw", "sepsis_raw.csv")
df = pd.read_csv(data_path)

# 2. Display first 3 rows
print("--- FIRST 3 ROWS OF DATA ---")
print(df.head(3))

# 3. Show dataset structural information
print("\n--- DATASET STRUCTURAL INFO ---")
print(df.info())
```

```python
# 4. Calculate exactly what percentage of data is missing for each column
print("--- TOP 10 COLUMNS WITH HIGHEST MISSING VALUES (%) ---")
missing_pct = (df.isnull().sum() / len(df)) * 100
print(missing_pct.sort_values(ascending=False).head(10))

# 5. Check if there are any completely identical rows
print("\n--- DUPLICATE ROWS COUNT ---")
print(f"Total duplicate rows found: {df.duplicated().sum()}")
```
